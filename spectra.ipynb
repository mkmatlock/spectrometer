```python
import numpy as np
import matplotlib
from matplotlib import pyplot as plt
import cv2
from IPython.display import Image, display
```

```python
cfl = cv2.imread('images/cfl.png')
old = cv2.imread('images/old.png')
old_rgb = cv2.cvtColor(old, cv2.COLOR_BGR2RGB)
old_int = cv2.cvtColor(old, cv2.COLOR_BGR2GRAY)
cfl_rgb = cv2.cvtColor(cfl, cv2.COLOR_BGR2RGB)
cfl_int = cv2.cvtColor(cfl, cv2.COLOR_BGR2GRAY)
cfl.shape
```

```python
fig, ax = plt.subplots(1,3, figsize=(15,5))
ax[0].imshow(cfl_rgb[:,:,0])
ax[0].set_title('Red Channel')
ax[1].imshow(cfl_rgb[:,:,1])
ax[1].set_title('Green Channel')
ax[2].imshow(cfl_rgb[:,:,2])
ax[2].set_title('Blue Channel')
plt.show()
```

```python
x1,y1,x2,y2 = 0, 1550, 3500, 1800
```

```python
plt.subplots(1,1,figsize=(10,8))
plt.imshow(cfl_rgb)
plt.vlines(x=[x1, x2], ymin=y1, ymax=y2, color='w')
plt.hlines(y=[y1, y2], xmin=x1, xmax=x2, color='w')
plt.show()
```

```python
plt.figure(figsize=(5,4))
x = np.arange(x1,x2)
y = cfl_int[y1:y2,x1:x2].mean(axis=0)
plt.xlabel('Wavelength (uncalibrated)')
plt.ylabel('Intensity')
plt.plot(x, y[::-1])
plt.xlim(0, 3050)
plt.show()
display(Image(filename='images/reference.png', width=450))
```
